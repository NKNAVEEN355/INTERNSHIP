```python
import pandas as pd
import numpy as np
filepa = r"C:\Users\DELL\Desktop\INTERNSHIP\kelunga_dummy_experts.xlsx"


df = pd.read_excel(filepa, sheet_name=0)

df['Skills'] = df['Skills'].apply(
    lambda x: x.split(',') if isinstance(x, str) else []
)

print(df[['Username','Skills']])


def find_experts_with_criteria(df, location=None, min_years=None, required_skills=None):

    result = df.copy()
    if location:
        result = result[result['Location'] == location]
    if min_years is not None:
        result = result[result['Years of Experience'] >= min_years]
    if required_skills:
        result = result[result['Skills'].apply(lambda x: all(skill in x for skill in required_skills))]
    return result


loc_anw = input("Do you want to filter by location? (yes/no): ")
if loc_anw.lower() == "yes":
    loc_input = input("Enter the location: ")
else:
    loc_input = None


years_input = input("Do you want to filter by minimum years of experience? (yes/no): ")
if years_input.lower() == "yes":
    min_years = int(input("Enter the minimum years of experience: "))
else:
    min_years = None


skills_input = input("Do you want to filter by required skills? (yes/no): ")
if skills_input.lower() == "yes":
    required_skills = [
        skill.strip() 
        for skill in input("Enter the required skills (comma-separated): ").split(",")
    ]
else:
    required_skills = None

recommended_experts = find_experts_with_criteria(
    df,
    location=loc_input,
    min_years=min_years,
    required_skills=required_skills
)

print("\nFiltered experts based on your criteria:")
print(recommended_experts)
```
