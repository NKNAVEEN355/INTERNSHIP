```python
import numpy as np

n = int(input('Enter the number of rolls: '))

   

rolls = np.random.randint(1, 7, size=n)

unique, counts = np.unique(rolls, return_counts=True)

print("\n--- Roll Results ---")
for vlue, count in zip(unique, counts):
    print('\nFace '+str(vlue)+' : '+str(count)+" times")


max = np.argmax(counts)
min=np.argmin(counts)
print('when we rolled '+str(n)+' times the max value shown is '+str(max)+'\n and min value is '+str(min))
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv" 
df = pd.read_csv(url)
print(df.head(5))


df_processed = df.copy()

df_processed['Age'] = df_processed['Age'].fillna(df_processed['Age'].mean())
df_processed['Fare'] = df_processed['Fare'].fillna(df_processed['Fare'].median())
print("\nMissing values in 'Age' and 'Fare' after filling data at missing :")
print(df_processed[['Age', 'Fare']].isnull().sum())



df_processed['Age_Z_score'] = (df_processed['Age'] - df_processed['Age'].mean()) / df_processed['Age'].std()

df_processed['Fare_Z_score'] = (df_processed['Fare'] - df_processed['Fare'].mean()) / df_processed['Fare'].std()


age_extreme_values = df_processed[df_processed['Age_Z_score'].abs() > 3]
print("\nExtreme values in Age (first 5 examples):")
print(age_extreme_values[['Age', 'Age_Z_score']].head())
print(f"Total Age extreme values identified: {len(age_extreme_values)}")


fare_extreme_values = df_processed[df_processed['Fare_Z_score'].abs() > 3]
print("\nExtreme values in Fare (first 5 examples):")
print(fare_extreme_values[['Fare', 'Fare_Z_score']].head())
print(f"Total Fare extreme values identified: {len(fare_extreme_values)}")

all_extreme_values = df_processed[(df_processed['Age_Z_score'].abs() > 3) | (df_processed['Fare_Z_score'].abs() > 3)]
print('\nTotal passengers identified as extreme values (in Age or Fare): '+str(len(all_extreme_values)))
print("Details of some 'odd passengers' (first 5 examples):")
print(all_extreme_values[['Age', 'Fare', 'Age_Z_score', 'Fare_Z_score']].head())



df_encoded = pd.get_dummies(df_processed, columns=['Sex'], dtype=int)

print(df_encoded)
```
